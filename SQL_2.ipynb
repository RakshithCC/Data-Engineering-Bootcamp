```sql
%sql
CREATE TABLE customers (
    customer_id INT ,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    city VARCHAR(50),
    email VARCHAR(100)
);
```

```sql
CREATE TABLE orders (
    order_id INT ,
    customer_id INT,
    order_date DATE,
    total_amount DECIMAL(10, 2)
);
```

```sql
CREATE TABLE order_details (
    order_detail_id INT ,
    order_id INT,
    customer_id INT,
    product_id INT,
    product_name VARCHAR(100),
    category VARCHAR(50),
    quantity INT,
    price_each DECIMAL(10,2),
    order_date DATE
);
```

```sql
select  e2.first_name, count(e1.emp_id) from emp e1 inner join emp e2 on e1.manager_id = e2.emp_id group by e2.first_name
```

```sql
SELECT e2.first_name AS manager, COUNT(e1.emp_id) AS team_size
FROM emp e1
INNER JOIN emp e2 ON e1.manager_id = e2.emp_id
GROUP BY e2.first_name;
```

```sql
select c.city, SUM(o.total_amount) as Total_Sales from customers c inner join orders o on c.customer_id = o.customer_id group by c.city
```

```sql
select c.first_name, c.last_name, sum(od.price_each * od.quantity)
from customers c join order_details od on c.customer_id = od.customer_id 
group by c.customer_id,c.first_name,c.last_name
having sum(od.price_each * od.quantity) > 2000
```

```sql
select c.city,o.order_id,MAX(od.price_each * od.quantity ) as sale_amount
from customers c join orders o on c.customer_id = o.customer_id
join order_details od on o.order_id = od.order_id
group by c.city, o.order_id
order by sale_amount desc
limit 1

```

```sql
WITH customer_orders AS(
  SELECT c.customer_id, c.first_name as Name, sum(od.quantity) as Total_Quantity FROM
  customers c join order_details od on c.customer_id = od.customer_id
  GROUP BY c.customer_id, c.first_name
) 
SELECT * FROM customer_orders WHERE Total_Quantity > 3
```

```sql
WITH Helper_table AS(
  SELECT c.first_name AS NAME ,o.order_id ,  count(o.order_id) AS total_orders
  FROM customers c JOIN orders o ON c.customer_id = o.customer_id GROUP by o.order_id,c.first_name
)
SELECT NAME, total_orders, 
CASE
  WHEN total_orders > 1 THEN 'Returning'
  ELSE 'New'
END as customer_type
from Helper_table
```

```sql
WITH customer_spend AS(
  SELECT c.customer_id,concat(c.first_name,' ',c.last_name) AS NAME ,  sum(od.price_each * od.quantity) AS total_spend
  FROM customers c JOIN orders o ON c.customer_id = o.customer_id JOIN order_details od ON o.order_id = od.order_id GROUP BY c.customer_id,c.first_name,c.last_name
)
SELECT NAME, total_spend,
CASE
  WHEN total_spend <1000 then 'Bronze'
  WHEN total_spend BETWEEN 1000 AND 2000 THEN 'Silver'
  ELSE 'Gold'
END as customer_segment
FROM customer_spend 
```
