```sql
CREATE TABLE Person  (
  customer_id VARCHAR(5),
  name VARCHAR(10),
  gender VARCHAR(6),
  city VARCHAR(10)
);
```

```sql
CREATE TABLE sales (
    sale_id INT PRIMARY KEY,
    customer_id VARCHAR(10),
    product VARCHAR(50),
    amount DECIMAL(10,2),
    sale_date DATE,
    region VARCHAR(20)
);
```

```sql
SELECT p.name, s.product, s.region from person p JOIN sales s ON p.customer_id = s.customer_id
```

```sql
SELECT p.name, count(s.customer_id)AS COUNT from person p join sales s on p.customer_id = s.customer_id GROUP BY s.customer_id, p.name 
```

```sql

 SELECT DISTINCT p.customer_id, p.name, p.gender
FROM person p
JOIN sales s ON p.customer_id = s.customer_id
WHERE s.amount > 1000
  AND p.gender = 'M';


```

```sql
SELECT s.region, SUM(s.amount) AS total_sales
FROM sales s
GROUP BY s.region
ORDER BY total_sales DESC
LIMIT 1;

```
